```python
import torch
import json
from pathlib import Path

hash = "e96207c2"


eval_path = Path(
    f"/home/jeclareyes/Documents/DL_Pipeline_AADT/DL_Pipeline_AADT/"
    f"outputs/runs/VI_Model/Synthetic04_{hash}/models/"
    f"Synthetic04_VI_Model_{hash}_eval_eval.pt"
)

out_path = eval_path.with_name("gradient_debug_first_epochs.json")

payload = torch.load(eval_path, map_location="cpu")
history = payload.get("epochs_history", {})

selected = {}

for epoch_key in sorted(history.keys(), key=lambda x: int(x))[:10]:
    record = history[epoch_key]

    selected[str(epoch_key)] = {
        "train_loss": record.get("train_loss"),
        "skipped_update": record.get("skipped_update"),
        "skip_reason": record.get("skip_reason"),
        "pre_clip_grad_norm": record.get("pre_clip_grad_norm"),
        "post_clip_grad_norm": record.get("post_clip_grad_norm"),
        "clip_applied": record.get("clip_applied"),
        "non_finite_gradients": record.get("non_finite_gradients"),
        "loss": record.get("loss", {}),
        "gradient_audit": record.get("gradient_audit", {}),
        "loss_gradient_audit": record.get("loss_gradient_audit", {}),
        "forward_physics_audit": record.get("forward_physics_audit", {}),
    }

out_path.write_text(
    json.dumps(selected, indent=2, ensure_ascii=False),
    encoding="utf-8",
)

print(f"Saved debug file to: {out_path}")
```
